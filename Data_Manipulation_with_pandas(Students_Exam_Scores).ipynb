# **Students Exam Scores Dataset Manipulation With Pandas**

```python
import pandas as pd
```

```python
df = pd.read_csv('/content/drive/MyDrive/Kaggle Datasets/Students Exam Scores/Original_data_with_more_rows.csv')
df.head()
```

## **Deleting Redundant Columns**

```python
df.columns
```

```python
drop_columns = ['Unnamed: 0', 'EthnicGroup', 'LunchType']
```

```python
df.drop(columns = drop_columns,inplace = True , errors= 'ignore')
df
```

```python
df.shape
```

## **Renaming The Columns**

```python
df.rename(columns = {"ParentEduc": "ParentEducation", "TestPrep": "TestPreparation" }, inplace= True) # you must put the argument inplace otherwise it won't change the columns name
df
```

```python
df.rename(columns={'Gender': 'Sex'}, inplace= True)
```

```python
df
```

## **Dropping Duplicate rows**

```python
df.duplicated()
```

```python
df.duplicated().value_counts()
```

```python
df.duplicated().sum()
```

### **Dropping Duplicated values**

```python
df.drop_duplicates(inplace = True)
```

```python
df
```

```python
df.shape
```

As we see the index_number remained unchanged. so, we have to reset the index

```python
df.reset_index(drop= True, inplace= True)
```

## **Removing NaN Values from The Dataset**

Checking if NaN value is present in the dataset.For that we have to use .isna() method, not .isnull()

```python
df.isna()
```

```python
df.isna().sum()
```

Since we didn't get any NaN values , we do not need to drop any NaN columns. If we had we would use .dropna(inplace = True).
