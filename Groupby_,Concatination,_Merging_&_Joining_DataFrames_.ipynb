# **Groupby & Aggregation**

GroupBy is one of the most powerful features in Pandas. It groups rows that share a value, then lets you calculate statistics per group - just

like a Pivot Table in Excel.

1. Basic GroupBy

2. Multiple aggregations with agg()


```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Titinic /titanic.csv")
```

## **1. Basic GroupBy**

## We can calculate statistical measurement by categorical data's group.

```python
mean_of_every_sex_group = df.groupby('Sex')['Age'].mean()
mean_of_every_sex_group
```

```python
median_of_every_sex_group = df.groupby('Sex')['Age'].median()
median_of_every_sex_group
```

```python
std_of_every_sex_group = df.groupby('Sex')['Age'].std()
std_of_every_sex_group
```

## **2. Multiple aggregations with agg()**

```python
Multiple_Measurement = df.groupby('Sex')['Age'].agg(["mean", "median", "max", "count"])
Multiple_Measurement
```

```python
Multiple_Measurement = df.groupby('Embarked')['Age'].agg(["mean", "median", "max", "count"])
Multiple_Measurement
```

## **Merging & Joining DataFrames**
Real projects have data spread across multiple tables. Merging combines two DataFrames using a common column - just like JOIN in SQL or VLOOKUP in Excel.

1. Inner join - keep only matches

2. Left join - keep all left rows

3. Right join - keep all Right rows
## `inner join`
pd.merge(df1, df2, on='id', how='inner')
## `left join`
pd.merge(df1, df2, on='id', how='left')
## `right join`
pd.merge(df1, df2, on='id', how='right')
## `outer join`
pd.merge(df1, df2, on='id', how='outer')

```python
df1 = pd.DataFrame( {
    "Name" : ["Rahat","Taha", "Abrar", "Sufian", "Emty"],
    "Age" : [ 23, 34, 32, 21, 12],
    "Scores" : [45, 35, 52, 23, 54]})
df2 = pd.DataFrame( {
    "Name" : ["Kawser","Sujon", "Babu", "Pias", "Faruk"],
    "Age" : [ 25, 37, 38, 29, 19],
    "Scores" : [55, 45, 72, 33, 84]})

display(df1) # to show dataframe as a table

print(df1)
```

## `inner join`

```python
pd.merge(df1, df2,  how = "inner")
```

## `left join`

```python
pd.merge(df1, df2, on = "Name", how = "left")
```

```python
pd.merge(df1, df2,  how = "left")
```

## `right join`

```python
pd.merge(df1, df2,  how = "right")
```

## `outer join`

```python
pd.merge(df1, df2,  how = "outer")
```
