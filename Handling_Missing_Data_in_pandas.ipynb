# **Handling Missing Data**

Real-world data is messy - it has missing values called NaN (Not a Number). Pandas gives you tools to find, fill, or remove these missing values before analysis.

1. Detect missing values
* df.isnull() # shows True/False
* df.isnull().sum() # count missing

2. Fill missing values fillna()
* df.fillna(0) # fill with 0


3. Drop missing ll missing rows. # df.dropna()

```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Titinic /titanic.csv")
```

## **`Detecting Missing Values  as True / False`**

```python
df.isnull()
```

## **`Detecting Total Missing Values  in every single columns`**

```python
df.isnull().sum()
```

## **`Filling missing values`**

**`For numeric  value`**

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df.isnull().sum()
```

**`For String  value`**

```python
df['Embarked'].mode()
```

```python
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode())
df.isnull().sum()
```

```python
df.info()
```

## **`Dropping All missing rows`**

```python
df.dropna() # see it removed all the missing rows . now we have 204 rows. previously we had 891 rows.
```
