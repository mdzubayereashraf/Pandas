# **Selection of Rows**

# **`loc → label based`**
1. df.loc[0] # row with index 0
2. df.loc[0:2, ['Name', 'Age']] #rows 0 to 2, selected cols

# **`iloc → index position based`**
1. df.iloc[0] # first row
2. df.iloc[0:3] # first 3 rows
3. df.iloc[:, 1] # all rows, 2nd col

## **`Selecting any particular row`**

```python
df.loc[0]
```

```python
df.loc[17]
```

## **`Selecting multiple particular rows from any position `**

```python
df.loc[4:9]
```

## **`Selecting a particular row and a column`**

```python
df.loc[9,'Name']
```

## **`Selecting multiple particular rows and a column`**

```python
df.loc[10: 15, 'Name']
```

## **`Selecting a particular rows and multiple columns`**

```python
df.loc[19,['Name','Age','Sex']]
```

## **`Selecting multiple particular rows and  columns simultaneously`**

```python
df.loc[10: 15, ['Name', 'Age']]
```

# **`iloc → index position based`**
## **`Selecting a single row by index number`**

```python
df.iloc[37]
```

## **`Selecting multiple row by index number from any range`**

```python
df.iloc[26:29]
```

## **`Selecting a single row  and column by index number`**

```python
df.iloc[43 , 3] # index 3 of column is Name
```

## **`Selecting multiple rows and columns  by index number`**

```python
df.iloc[3:7, 1: 4]
```
