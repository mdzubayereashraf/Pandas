# **Filtering Data**

Filtering means showing only the rows that match a condition like a WHERE clause in SQL or a filter in Excel. You write a condition, and Pandas returns only the True rows.

1. Filter with one condition. # `df[df['Age'] > 30]`

2. Filter with Multiple conditions. # `df[(df['Age'] > 35) & (df['Fare'] < 50.000)]`

3. isin() - filter from a list. # `df[df["Name"].isin(names)]`

4. query() - filter using plain text



## **`1. Filter with one condition`**

```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Titinic /titanic.csv")

df[df['Age'] > 30]
```

```python
df[df['Fare'] > 50.000]
```

## **`2. Filter with Multiple conditions`**

```python
df[(df['Age'] > 35) & (df['Fare'] < 50.000)]
```

## **`3. isin() - filter from a list`**

```python
names = ["Rice, Mrs. William (Margaret Norton)", "Bystrom, Mrs. (Karolina)", "Hansen, Mr. Claus Peter"]
df[df["Name"].isin(names)]
```

## **`4. query() - filter using plain text`**

```python
df.query("Age > 40")
```

```python
df.query("Age > 40 & Fare > 50")
```

# **Adding & Modifying Columns**

In data analysis, you often need to create new columns from existing ones - like calculating a grade from a score, or combining first and last name into a full name.


## **`Adding new column`**

```python
df["Half Price"] = df['Fare'] * .50
df.head()
```

## **`Modify an existing column`**

```python
df["Family"] = df['SibSp'] + df['Parch']
df.head()
```

## **`Dropping  a column`**

```python
df.drop('Cabin', axis= 1)
```

## **`Dropping  a row `**

```python
df.drop(0, axis= 0)
```

# **Renaming columns**

```python
df.rename( columns={'Fare' : 'Price'}, inplace= True)
df.head()
```

# **Sorting**

## **`With Ascending Order`**

```python
df.sort_values('Price', ascending= True)
```

## **`With Descending Order``**

```python
df.sort_values('Price', ascending= False)
```

# **`Sorting Multiple Columns`**

```python
df.sort_values(["Family", "Age"], ascending= [True , False]) # Need to put multiple columns in a list. Family = True , Age = False
```
