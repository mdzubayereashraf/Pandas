#**Introduction**

Pandas is a Python library built on top of NumPy, designed to work with structured data like tables and spreadsheets. Just like NumPy gave us powerful arrays, Pandas gives us powerful tables called DataFrames.

**`What is Pandas?`**

Pandas = Panel Data. It lets you load, clean, filter, and analyze data in rows and columns just like Excel, but with Python code.

**`Import Pandas`**

We always import pandas as pd this is a universal convention in the Python world.









## **`Series - 1D Data`**

A Series is like a single column of a table a list of values with labels called index. Think of it as a NumPy array with named rows.

**`Creating a Series from a list`**

```python
import pandas as pd

scores = pd.Series([34, 53, 24, 42, 25, 13, 35, 16 ,45])
scores
```

```python
scores = pd.Series([34, 53, 24, 42, 25, 13, 35, 16 ,45], name ="Numbers")
scores
```

## **`Series with custom index`**

```python
scores = pd.Series([34, 53, 24, 42, 25, 13, 35, 16 ,45], name = " Numbers", index = [1,2,3,4,5,6,7,8,9])
scores
```

```python
scores = pd.Series([34, 53, 24, 42, 25, 13, 35, 16 ,45],
                   name = " Numbers", index = ["Taki", "Yamin", "Ashraf", "Abir", "Akash"," Palak", "Pakhi", "Rahat", "Faruk"])
scores
```

## **`Series from a dictionary`**

```python
dic = {
    "Rafi" : 45,
    "Kafi" : 65,
    "Naim" : 75,
    "Raki" : 45
}
scores = pd.Series(dic, name = "Numbers")
scores
```

## **`DataFrame - 2D Table`**

A DataFrame is the heart of Pandas. It is a table with rows and columns they form a DataFrame. exactly like a spreadsheet. Each column is a Series, and together







## **`Creating a DataFrame from a dictionary`**

```python
dic = {
    "Name" : ["Rahat","Taha", "Abrar", "Sufian", "Emty"],
    "Age" : [ 23, 34, 32, 21, 12],
    "Scores" : [45, 35, 52, 23, 54]
}
df = pd.DataFrame(dic)
df
```

## **`Explore the DataFrame`**

```python
dic = {
    "Name" : ["Rahat","Taha", "Abrar", "Sufian", "Emty"],
    "Age" : [ 23, 34, 32, 21, 12],
    "Scores" : [45, 35, 52, 23, 54]
}
df = pd.DataFrame(dic)

print(df.shape)
print(df.dtypes)
print(df.size)


```
