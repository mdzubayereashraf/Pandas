# **Reading, Viewing, and, Saving Data**

In real projects, data comes from files code. not code. Pandas can read CSV files, Excel files, JSON, and even SQL databases with a single line of code.

**`Functions:`**
* pd.read_csv('data.csv')
* pd.read_excel('data.xlsx')
* pd.read_json('data.json')
* pd.read_sql('query', connection)

**`To View Data:`**
1. df.head() # first 5 rows
2. df.tail() # last 5 rows
3. df.sample(3) # random 3 rows
4.  df.shape # (rows, cols)
5. df.columns # column names
6. df.index # index
7. df.size # to count total values
8. df.info() # info about df
9. df.describe() # summary stats

**`To Save data as different extensions:`**
* df.to_csv('out.csv', index=False)
* df.to_excel('out.xlsx', index=False)
* df.to_json('out.json')
